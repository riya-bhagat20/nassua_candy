```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
import mysql.connector
```

```python
url = "https://www.nassaucandy.com/"
soup = BeautifulSoup(requests.get(url).text, 'html.parser')

# ── Saari unique tags dhundo ──
print("=== TAGS ===")
all_tags = set([tag.name for tag in soup.find_all()])
print(all_tags)

# ── Saari CSS classes dhundo ──
print("\n=== CLASSES ===")
for tag in soup.find_all(class_=True):
    print(f"Tag: <{tag.name}>  Class: {tag['class']}")

# ── Saara visible text dhundo ──
print("\n=== TEXT CONTENT ===")
for tag in soup.find_all(['h1','h2','h3','p','span','td','a']):
    text = tag.get_text(strip=True)
    if text:
        print(f"<{tag.name}>: {text[:60]}")

# ── Saare links aur images ──
print("\n=== LINKS ===")
for a in soup.find_all('a', href=True)[:10]:
    print(a['href'])

print("\n=== IMAGES ===")
for img in soup.find_all('img')[:5]:
    print(img.get('src'))
```
